```python
# BiLSTM-dynamic-att-LSTM on ACTI corpus (PyTorch implementation)
# ---------------------------------------------------------------
# This script trains a BiLSTM-dynamic-att-LSTM joint tagger on the ACTI dataset
# and plots Precision / Recall / F1 over epochs.
#
# You MUST set the paths below (TAG_DIC_PATH, TRAIN_PATH, DEV_PATH)
# to point to your local copies of tag.dic, train.txt and dev.txt.

# Optional (for Colab / Jupyter):
# !pip install -q seqeval scikit-learn

import os
from collections import Counter

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    classification_report,
)

import matplotlib.pyplot as plt   # <-- for graphs

# ============================================
# 1. Paths to ACTI files  (CHANGE THESE!)
# ============================================
TAG_DIC_PATH = "model/BiLSTM-dynamic-att-LSTM/data1/tag.dic"
TRAIN_PATH   = "model/BiLSTM-dynamic-att-LSTM/data1/train.txt"
DEV_PATH     = "model/BiLSTM-dynamic-att-LSTM/data1/dev.txt"  # used as test set here


# ============================================
# 2. Utility: load tag dictionary
# ============================================
def load_labels(path):
    with open(path, encoding="utf-8") as f:
        labs = [line.strip() for line in f if line.strip()]
    return labs


labels = load_labels(TAG_DIC_PATH)
num_labels = len(labels)
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for i, lab in enumerate(labels)}

print("Number of labels:", num_labels)


# ============================================
# 3. Load ACTI train/dev (tokens + labels)
#    Each line: "token1 token2 ...\tlabel1 label2 ..."
# ============================================
def load_acti_file(path):
    sentences = []
    tags = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue
            parts = line.split("\t")
            if len(parts) != 2:
                # skip malformed
                continue
            sent_str, tag_str = parts
            tokens = sent_str.split()
            labs   = tag_str.split()
            assert len(tokens) == len(labs), f"Token/label mismatch in {path}"
            sentences.append(tokens)
            tags.append(labs)
    return sentences, tags


train_tokens, train_labels = load_acti_file(TRAIN_PATH)
dev_tokens,   dev_labels   = load_acti_file(DEV_PATH)

print("Train sentences:", len(train_tokens))
print("Dev sentences:  ", len(dev_tokens))


# ============================================
# 4. Build word & char vocab from training data
# ============================================
word_counter = Counter()
char_counter = Counter()

for sent in train_tokens:
    word_counter.update(sent)
    for tok in sent:
        char_counter.update(list(tok))

# word vocab
word_vocab = {"<PAD>": 0, "<UNK>": 1}
for w in word_counter.keys():
    if w not in word_vocab:
        word_vocab[w] = len(word_vocab)

# char vocab
char_vocab = {"<PAD>": 0, "<UNK>": 1}
for ch in char_counter.keys():
    if ch not in char_vocab:
        char_vocab[ch] = len(char_vocab)

vocab_size = len(word_vocab)
char_vocab_size = len(char_vocab)
print("Word vocab size:", vocab_size)
print("Char vocab size:", char_vocab_size)


# ============================================
# 5. Dataset + collate_fn (with chars)
# ============================================
class ActiDataset(Dataset):
    def __init__(self, sentences, labels, word_vocab, char_vocab, label2id):
        self.sentences = sentences
        self.labels = labels
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.label2id = label2id

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx):
        tokens = self.sentences[idx]
        labs   = self.labels[idx]

        # word ids
        token_ids = [self.word_vocab.get(t, self.word_vocab["<UNK>"]) for t in tokens]
        # label ids
        label_ids = [self.label2id[l] for l in labs]
        # char ids (list of 1D tensors, one per token)
        char_ids = []
        for tok in tokens:
            chars = [self.char_vocab.get(ch, self.char_vocab["<UNK>"]) for ch in tok]
            char_ids.append(torch.tensor(chars, dtype=torch.long))

        return {
            "token_ids": torch.tensor(token_ids, dtype=torch.long),
            "char_ids": char_ids,  # list of [len(token)]
            "labels": torch.tensor(label_ids, dtype=torch.long),
        }


def collate_fn(batch):
    """
    Pads:
      - word sequences to max_seq_len
      - char sequences to max_word_len
      - labels with -1 (ignore_index)
    Returns:
      input_ids: [B, T]
      char_ids:  [B, T, C]
      labels:    [B, T]
      mask:      [B, T] bool (True where real token)
    """
    token_seqs = [item["token_ids"] for item in batch]
    char_seqs  = [item["char_ids"] for item in batch]  # list of list[tensor(word_len)]
    label_seqs = [item["labels"] for item in batch]

    batch_size = len(batch)
    lengths = [len(seq) for seq in token_seqs]
    max_len = max(lengths)

    # maximum word length over the whole batch
    max_word_len = 1
    for sent_chars in char_seqs:
        for wchars in sent_chars:
            if len(wchars) > max_word_len:
                max_word_len = len(wchars)

    pad_token_id = word_vocab["<PAD>"]
    pad_char_id = char_vocab["<PAD>"]
    pad_label = -1  # ignore index

    padded_tokens = []
    padded_labels = []
    padded_masks  = []
    padded_chars  = []

    for tokens, sent_chars, labels in zip(token_seqs, char_seqs, label_seqs):
        seq_len = len(tokens)
        pad_len = max_len - seq_len

        # words
        tokens_padded = torch.cat(
            [tokens, torch.full((pad_len,), pad_token_id, dtype=torch.long)]
        )
        labels_padded = torch.cat(
            [labels, torch.full((pad_len,), pad_label, dtype=torch.long)]
        )
        mask = torch.cat(
            [
                torch.ones(seq_len, dtype=torch.bool),
                torch.zeros(pad_len, dtype=torch.bool),
            ]
        )

        # chars -> [T, max_word_len]
        word_char_tensors = []
        for i in range(max_len):
            if i < seq_len:
                wchars = sent_chars[i]
                wlen = len(wchars)
                if wlen < max_word_len:
                    wchars = torch.cat(
                        [
                            wchars,
                            torch.full(
                                (max_word_len - wlen,),
                                pad_char_id,
                                dtype=torch.long,
                            ),
                        ]
                    )
                else:
                    wchars = wchars[:max_word_len]
            else:
                # padded word -> all PAD chars
                wchars = torch.full((max_word_len,), pad_char_id, dtype=torch.long)
            word_char_tensors.append(wchars)
        char_tensor = torch.stack(word_char_tensors, dim=0)  # [T, C]

        padded_tokens.append(tokens_padded)
        padded_labels.append(labels_padded)
        padded_masks.append(mask)
        padded_chars.append(char_tensor)

    batch_input_ids = torch.stack(padded_tokens, dim=0)  # [B, T]
    batch_labels    = torch.stack(padded_labels, dim=0)  # [B, T]
    batch_masks     = torch.stack(padded_masks, dim=0)   # [B, T]
    batch_char_ids  = torch.stack(padded_chars, dim=0)   # [B, T, C]

    return {
        "input_ids": batch_input_ids,
        "char_ids": batch_char_ids,
        "labels": batch_labels,
        "mask": batch_masks,
    }


train_dataset = ActiDataset(train_tokens, train_labels, word_vocab, char_vocab, label2id)
dev_dataset   = ActiDataset(dev_tokens,   dev_labels,   word_vocab, char_vocab, label2id)

train_loader = DataLoader(
    train_dataset, batch_size=16, shuffle=True, collate_fn=collate_fn
)
dev_loader = DataLoader(
    dev_dataset, batch_size=16, shuffle=False, collate_fn=collate_fn
)

print("DataLoaders ready.")


# ============================================
# 6. BiLSTM–Dynamic-Attention–LSTM model
#    (word2vec + char-CNN + BiLSTM + self-attn + GRU + LSTM)
# ============================================
class BiLSTMDynamicAttLSTM(nn.Module):
    def __init__(
        self,
        vocab_size,
        char_vocab_size,
        num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,
    ):
        super().__init__()
        self.num_labels = num_labels
        self.word_emb_dim = word_emb_dim
        self.char_emb_dim = char_emb_dim
        self.cnn_filters = cnn_filters
        self.bilstm_dim = bilstm_dim
        self.lstm_dim = lstm_dim

        # 1) Embeddings
        self.word_embedding = nn.Embedding(vocab_size, word_emb_dim, padding_idx=0)
        self.char_embedding = nn.Embedding(
            char_vocab_size, char_emb_dim, padding_idx=0
        )

        # character-level CNN
        self.char_cnn = nn.Conv1d(
            in_channels=char_emb_dim,
            out_channels=cnn_filters,
            kernel_size=cnn_kernel_size,
            padding=cnn_kernel_size // 2,
            bias=True,
        )

        # 2) Encoder BiLSTM
        self.encoder = nn.LSTM(
            input_size=word_emb_dim + cnn_filters,
            hidden_size=bilstm_dim // 2,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        # 3) Self-attention
        self.attn = nn.MultiheadAttention(
            embed_dim=bilstm_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        # 4) Dynamic attention gating: Ws [h_t, a_t] -> gamma_t
        self.W_s = nn.Linear(bilstm_dim * 2, bilstm_dim * 2)

        # GRU after dynamic attention
        self.gru = nn.GRU(
            input_size=bilstm_dim * 2,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        # 5) Decoder LSTM
        self.decoder = nn.LSTM(
            input_size=lstm_dim,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(lstm_dim, num_labels)

        # optional "bias weight" hyperparameter (initialise to 10.0 like paper)
        self.bias_weight = nn.Parameter(torch.tensor(10.0))

    def forward(self, input_ids, char_ids, mask, labels=None):
        """
        input_ids: [B, T]
        char_ids:  [B, T, C]
        mask:      [B, T] bool
        labels:    [B, T] or None
        """
        B, T = input_ids.shape
        _, _, C = char_ids.shape

        # --- Word embeddings ---
        word_emb = self.word_embedding(input_ids)  # [B, T, D_w]

        # --- Char-CNN embeddings ---
        # char_ids: [B, T, C] -> [B*T, C]
        char_ids_flat = char_ids.view(B * T, C)
        # [B*T, C, D_c]
        char_emb = self.char_embedding(char_ids_flat)
        # conv1d expects [N, C_in, L] -> transpose
        char_emb = char_emb.transpose(1, 2)  # [B*T, D_c, C]
        char_conv = self.char_cnn(char_emb)  # [B*T, F, C]
        char_conv = torch.relu(char_conv)
        # max pool over time dimension C
        char_rep = torch.max(char_conv, dim=2).values  # [B*T, F]
        char_rep = char_rep.view(B, T, self.cnn_filters)  # [B, T, F]

        # zero out char reps for padding positions
        mask_float = mask.unsqueeze(-1).float()
        char_rep = char_rep * mask_float

        # concat word & char features
        enc_input = torch.cat([word_emb, char_rep], dim=-1)  # [B, T, D_w+F]

        # --- BiLSTM encoder ---
        enc_out, _ = self.encoder(enc_input)  # [B, T, bilstm_dim]

        # --- Self-attention ---
        key_padding_mask = ~mask  # True where pad
        attn_out, _ = self.attn(
            enc_out, enc_out, enc_out, key_padding_mask=key_padding_mask
        )  # [B, T, bilstm_dim]

        # --- Dynamic attention gating ---
        # concat h_t and a_t
        concat_ha = torch.cat([enc_out, attn_out], dim=-1)  # [B, T, 2*bilstm_dim]
        # gamma_t = sigmoid(W_s [h_t, a_t] + bias_weight)
        gamma = torch.sigmoid(self.W_s(concat_ha) + self.bias_weight)  # [B, T, 2*bilstm_dim]
        # xi_t = gamma_t * [h_t, a_t]
        xi = gamma * concat_ha  # [B, T, 2*bilstm_dim]

        # --- GRU over dynamic attention outputs ---
        gru_out, _ = self.gru(xi)  # [B, T, lstm_dim]

        # --- Decoder LSTM ---
        dec_out, _ = self.decoder(gru_out)  # [B, T, lstm_dim]

        logits = self.classifier(self.dropout(dec_out))  # [B, T, num_labels]

        if labels is not None:
            # Flatten for loss
            loss_fn = nn.CrossEntropyLoss(ignore_index=-1)
            loss = loss_fn(logits.view(-1, self.num_labels), labels.view(-1))
            return loss
        else:
            preds = logits.argmax(dim=-1)  # [B, T]
            return preds


# ============================================
# 7. Evaluation: seqeval entity-level metrics
# ============================================
def evaluate(model, data_loader, device, num_epochs=0):
    model.eval()
    seqeval_true = []
    seqeval_pred = []
    total_loss = 0.0
    steps = 0

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            loss = model(input_ids, char_ids, mask, labels=labels)
            total_loss += loss.item()
            steps += 1

            preds = model(input_ids, char_ids, mask, labels=None)  # [B, T]
            B, T = input_ids.shape

            for i in range(B):
                true_seq = []
                pred_seq = []
                for j in range(T):
                    if not mask[i, j]:
                        continue
                    true_lab_id = labels[i, j].item()
                    if true_lab_id == -1:
                        continue
                    pred_lab_id = preds[i, j].item()
                    true_seq.append(id2label[true_lab_id])
                    pred_seq.append(id2label[pred_lab_id])
                if len(true_seq) > 0:
                    seqeval_true.append(true_seq)
                    seqeval_pred.append(pred_seq)

    avg_loss = total_loss / steps if steps > 0 else 0.0

    ent_precision = precision_score(seqeval_true, seqeval_pred)
    ent_recall    = recall_score(seqeval_true, seqeval_pred)
    ent_f1        = f1_score(seqeval_true, seqeval_pred)
    ent_accuracy  = accuracy_score(seqeval_true, seqeval_pred)

    eval_results = {
        "eval_loss": float(avg_loss),
        "eval_precision": float(ent_precision),
        "eval_recall": float(ent_recall),
        "eval_f1": float(ent_f1),
        "eval_accuracy": float(ent_accuracy),
        "epoch": float(num_epochs),
    }

    print("Eval results:", eval_results)
    print("\nClassification report:\n")
    print(classification_report(seqeval_true, seqeval_pred))

    return eval_results


# ============================================
# 8. Training loop + metric graphs
# ============================================
def main():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print("Using device:", device)

    model = BiLSTMDynamicAttLSTM(
        vocab_size=vocab_size,
        char_vocab_size=char_vocab_size,
        num_labels=num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,
    ).to(device)

    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)  # 0.001 as in Table 6

    num_epochs = 10  # Increase (e.g., 50–100) for closer-to-paper results

    # ---- history for plotting ----
    epochs_list    = []
    precision_list = []
    recall_list    = []
    f1_list        = []
    train_loss_list = []

    for epoch in range(1, num_epochs + 1):
        model.train()
        total_loss = 0.0

        for batch in train_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            optimizer.zero_grad()
            loss = model(input_ids, char_ids, mask, labels=labels)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        avg_train_loss = total_loss / len(train_loader)
        train_loss_list.append(avg_train_loss)
        print(f"Epoch {epoch}: train loss = {avg_train_loss:.4f}")

        # ---- evaluate on dev after each epoch ----
        eval_results = evaluate(model, dev_loader, device, num_epochs=epoch)

        # store metrics for plotting
        epochs_list.append(epoch)
        precision_list.append(eval_results["eval_precision"])
        recall_list.append(eval_results["eval_recall"])
        f1_list.append(eval_results["eval_f1"])

    # =====================================
    #  Plot Precision, Recall, F1 vs Epoch
    # =====================================
    plt.figure()
    plt.plot(epochs_list, precision_list, label="Precision")
    plt.plot(epochs_list, recall_list,    label="Recall")
    plt.plot(epochs_list, f1_list,        label="F1-score")
    plt.xlabel("Epoch")
    plt.ylabel("Score")
    plt.title("Dev Precision / Recall / F1 over Epochs")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()   # or plt.savefig("metrics_curve.png")

    # (Optional) plot training loss
    plt.figure()
    plt.plot(epochs_list, train_loss_list, label="Train loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training Loss over Epochs")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    print("Final evaluation on dev set:")
    evaluate(model, dev_loader, device, num_epochs=num_epochs)


if __name__ == "__main__":
    main()
```

```python
# BiLSTM–Dynamic-Attention–LSTM on ACTI corpus (PyTorch) + 10 independent iterations (seeds)
# ----------------------------------------------------------------------------------------
# - Uses tag.dic, train.txt, dev.txt (ACTI format)
# - Word embedding + char-CNN + BiLSTM + self-attn + dynamic gating + GRU + LSTM
# - seqeval for Precision / Recall / F1 / Accuracy
# - Runs 10 independent runs (different seeds), retraining from scratch each run
#
# Colab:
#   !pip -q install seqeval

import random
from collections import Counter

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    classification_report,
)

# ============================================
# 1) Paths to ACTI files (CHANGE THESE!)
# ============================================
TAG_DIC_PATH = "model/BiLSTM-dynamic-att-LSTM/data1/tag.dic"
TRAIN_PATH   = "model/BiLSTM-dynamic-att-LSTM/data1/train.txt"
DEV_PATH     = "model/BiLSTM-dynamic-att-LSTM/data1/dev.txt"


# ============================================
# 2) Reproducibility helper
# ============================================
def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================
# 3) Load labels
# ============================================
def load_labels(path):
    with open(path, encoding="utf-8") as f:
        labs = [line.strip() for line in f if line.strip()]
    return labs

labels = load_labels(TAG_DIC_PATH)
num_labels = len(labels)
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for i, lab in enumerate(labels)}

print("Number of labels:", num_labels)


# ============================================
# 4) Load ACTI train/dev (tokens + labels)
# Each line: "tok tok ...\tLAB LAB ..."
# ============================================
def load_acti_file(path):
    sentences, tags = [], []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue
            parts = line.split("\t")
            if len(parts) != 2:
                continue
            sent_str, tag_str = parts
            tokens = sent_str.split()
            labs = tag_str.split()
            if len(tokens) != len(labs):
                raise ValueError(f"Token/label mismatch in {path}")
            if len(tokens) == 0:
                continue
            sentences.append(tokens)
            tags.append(labs)
    return sentences, tags

train_tokens, train_labels = load_acti_file(TRAIN_PATH)
dev_tokens,   dev_labels   = load_acti_file(DEV_PATH)

print("Train sentences:", len(train_tokens))
print("Dev sentences:  ", len(dev_tokens))


# ============================================
# 5) Build word & char vocab from training data
# ============================================
word_counter = Counter()
char_counter = Counter()

for sent in train_tokens:
    word_counter.update(sent)
    for tok in sent:
        char_counter.update(list(tok))

word_vocab = {"<PAD>": 0, "<UNK>": 1}
for w in word_counter.keys():
    if w not in word_vocab:
        word_vocab[w] = len(word_vocab)

char_vocab = {"<PAD>": 0, "<UNK>": 1}
for ch in char_counter.keys():
    if ch not in char_vocab:
        char_vocab[ch] = len(char_vocab)

vocab_size = len(word_vocab)
char_vocab_size = len(char_vocab)
print("Word vocab size:", vocab_size)
print("Char vocab size:", char_vocab_size)


# ============================================
# 6) Dataset + collate_fn (with chars)
# ============================================
class ActiDataset(Dataset):
    def __init__(self, sentences, labels, word_vocab, char_vocab, label2id):
        self.sentences = sentences
        self.labels = labels
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.label2id = label2id

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx):
        tokens = self.sentences[idx]
        labs = self.labels[idx]

        token_ids = [self.word_vocab.get(t, self.word_vocab["<UNK>"]) for t in tokens]
        label_ids = [self.label2id[l] for l in labs]

        # list[tensor(word_len)]
        char_ids = []
        for tok in tokens:
            chars = [self.char_vocab.get(ch, self.char_vocab["<UNK>"]) for ch in tok]
            char_ids.append(torch.tensor(chars, dtype=torch.long))

        return {
            "token_ids": torch.tensor(token_ids, dtype=torch.long),
            "char_ids": char_ids,
            "labels": torch.tensor(label_ids, dtype=torch.long),
        }

def make_collate_fn(word_vocab, char_vocab, pad_label_id=-1):
    pad_token_id = word_vocab["<PAD>"]
    pad_char_id = char_vocab["<PAD>"]

    def collate_fn(batch):
        token_seqs = [item["token_ids"] for item in batch]
        char_seqs  = [item["char_ids"] for item in batch]
        label_seqs = [item["labels"] for item in batch]

        lengths = [len(seq) for seq in token_seqs]
        max_len = max(lengths)

        max_word_len = 1
        for sent_chars in char_seqs:
            for wchars in sent_chars:
                if len(wchars) > max_word_len:
                    max_word_len = len(wchars)

        padded_tokens, padded_labels, padded_masks, padded_chars = [], [], [], []

        for tokens, sent_chars, labels_t in zip(token_seqs, char_seqs, label_seqs):
            seq_len = len(tokens)
            pad_len = max_len - seq_len

            tokens_padded = torch.cat([tokens, torch.full((pad_len,), pad_token_id, dtype=torch.long)])
            labels_padded = torch.cat([labels_t, torch.full((pad_len,), pad_label_id, dtype=torch.long)])

            mask = torch.cat([
                torch.ones(seq_len, dtype=torch.bool),
                torch.zeros(pad_len, dtype=torch.bool),
            ])

            # chars: [max_len, max_word_len]
            word_char_tensors = []
            for i in range(max_len):
                if i < seq_len:
                    wchars = sent_chars[i]
                    wlen = len(wchars)
                    if wlen < max_word_len:
                        wchars = torch.cat([wchars, torch.full((max_word_len - wlen,), pad_char_id, dtype=torch.long)])
                    else:
                        wchars = wchars[:max_word_len]
                else:
                    wchars = torch.full((max_word_len,), pad_char_id, dtype=torch.long)
                word_char_tensors.append(wchars)

            char_tensor = torch.stack(word_char_tensors, dim=0)  # [T, C]

            padded_tokens.append(tokens_padded)
            padded_labels.append(labels_padded)
            padded_masks.append(mask)
            padded_chars.append(char_tensor)

        return {
            "input_ids": torch.stack(padded_tokens, dim=0),  # [B, T]
            "char_ids":  torch.stack(padded_chars,  dim=0),  # [B, T, C]
            "labels":    torch.stack(padded_labels, dim=0),  # [B, T]
            "mask":      torch.stack(padded_masks,  dim=0),  # [B, T]
        }

    return collate_fn


train_dataset = ActiDataset(train_tokens, train_labels, word_vocab, char_vocab, label2id)
dev_dataset   = ActiDataset(dev_tokens,   dev_labels,   word_vocab, char_vocab, label2id)


# ============================================
# 7) BiLSTM–Dynamic-Attention–LSTM model
# ============================================
class BiLSTMDynamicAttLSTM(nn.Module):
    def __init__(
        self,
        vocab_size,
        char_vocab_size,
        num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,
    ):
        super().__init__()
        self.num_labels = num_labels
        self.word_emb_dim = word_emb_dim
        self.char_emb_dim = char_emb_dim
        self.cnn_filters = cnn_filters
        self.bilstm_dim = bilstm_dim
        self.lstm_dim = lstm_dim

        self.word_embedding = nn.Embedding(vocab_size, word_emb_dim, padding_idx=0)
        self.char_embedding = nn.Embedding(char_vocab_size, char_emb_dim, padding_idx=0)

        self.char_cnn = nn.Conv1d(
            in_channels=char_emb_dim,
            out_channels=cnn_filters,
            kernel_size=cnn_kernel_size,
            padding=cnn_kernel_size // 2,
            bias=True,
        )

        self.encoder = nn.LSTM(
            input_size=word_emb_dim + cnn_filters,
            hidden_size=bilstm_dim // 2,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        self.attn = nn.MultiheadAttention(
            embed_dim=bilstm_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        self.W_s = nn.Linear(bilstm_dim * 2, bilstm_dim * 2)

        self.gru = nn.GRU(
            input_size=bilstm_dim * 2,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.decoder = nn.LSTM(
            input_size=lstm_dim,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(lstm_dim, num_labels)

        # as in the paper
        self.bias_weight = nn.Parameter(torch.tensor(10.0))

    def forward(self, input_ids, char_ids, mask, labels=None):
        B, T = input_ids.shape
        _, _, C = char_ids.shape

        word_emb = self.word_embedding(input_ids)  # [B, T, D_w]

        # char-cnn
        char_ids_flat = char_ids.view(B * T, C)         # [B*T, C]
        char_emb = self.char_embedding(char_ids_flat)   # [B*T, C, D_c]
        char_emb = char_emb.transpose(1, 2)             # [B*T, D_c, C]
        char_conv = torch.relu(self.char_cnn(char_emb)) # [B*T, F, C]
        char_rep = torch.max(char_conv, dim=2).values   # [B*T, F]
        char_rep = char_rep.view(B, T, self.cnn_filters)# [B, T, F]

        # zero out for padding positions
        char_rep = char_rep * mask.unsqueeze(-1).float()

        enc_input = torch.cat([word_emb, char_rep], dim=-1)  # [B, T, D_w+F]
        enc_out, _ = self.encoder(enc_input)                 # [B, T, H]

        key_padding_mask = ~mask
        attn_out, _ = self.attn(enc_out, enc_out, enc_out, key_padding_mask=key_padding_mask)  # [B, T, H]

        concat_ha = torch.cat([enc_out, attn_out], dim=-1)                    # [B, T, 2H]
        gamma = torch.sigmoid(self.W_s(concat_ha) + self.bias_weight)         # [B, T, 2H]
        xi = gamma * concat_ha                                                # [B, T, 2H]

        gru_out, _ = self.gru(xi)                                             # [B, T, lstm_dim]
        dec_out, _ = self.decoder(gru_out)                                    # [B, T, lstm_dim]
        logits = self.classifier(self.dropout(dec_out))                       # [B, T, num_labels]

        if labels is not None:
            loss_fn = nn.CrossEntropyLoss(ignore_index=-1)
            return loss_fn(logits.view(-1, self.num_labels), labels.view(-1))
        return logits.argmax(dim=-1)  # [B, T]


# ============================================
# 8) Evaluation (seqeval)
# ============================================
def evaluate(model, data_loader, device, print_report=False):
    model.eval()
    seqeval_true, seqeval_pred = [], []
    total_loss, steps = 0.0, 0

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            loss = model(input_ids, char_ids, mask, labels=labels)
            total_loss += loss.item()
            steps += 1

            preds = model(input_ids, char_ids, mask, labels=None)  # [B, T]
            B, T = input_ids.shape

            for i in range(B):
                true_seq, pred_seq = [], []
                for j in range(T):
                    if not mask[i, j]:
                        continue
                    true_lab_id = int(labels[i, j].item())
                    if true_lab_id == -1:
                        continue
                    pred_lab_id = int(preds[i, j].item())
                    true_seq.append(id2label[true_lab_id])
                    pred_seq.append(id2label[pred_lab_id])

                if true_seq:
                    seqeval_true.append(true_seq)
                    seqeval_pred.append(pred_seq)

    avg_loss = total_loss / steps if steps > 0 else 0.0
    p = precision_score(seqeval_true, seqeval_pred)
    r = recall_score(seqeval_true, seqeval_pred)
    f = f1_score(seqeval_true, seqeval_pred)
    a = accuracy_score(seqeval_true, seqeval_pred)

    results = {
        "eval_loss": float(avg_loss),
        "eval_precision": float(p),
        "eval_recall": float(r),
        "eval_f1": float(f),
        "eval_accuracy": float(a),
    }

    if print_report:
        print("\nClassification report:\n")
        print(classification_report(seqeval_true, seqeval_pred))

    return results


# ============================================
# 9) One run (one seed) training
# ============================================
def run_one_iteration(
    seed: int,
    num_epochs: int = 10,
    batch_size: int = 16,
    lr: float = 1e-3,
    grad_clip: float = 5.0,
    use_best_dev: bool = True,   # True: return best dev F1 across epochs; False: final epoch
    print_each_epoch: bool = True,
):
    set_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    collate_fn = make_collate_fn(word_vocab, char_vocab, pad_label_id=-1)

    # deterministic shuffle order for this seed
    g = torch.Generator()
    g.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=collate_fn,
        generator=g,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )
    dev_loader = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=collate_fn,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    model = BiLSTMDynamicAttLSTM(
        vocab_size=vocab_size,
        char_vocab_size=char_vocab_size,
        num_labels=num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,
    ).to(device)

    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    best_res = None
    last_res = None

    for epoch in range(1, num_epochs + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            optimizer.zero_grad()
            loss = model(input_ids, char_ids, mask, labels=labels)
            loss.backward()

            if grad_clip is not None:
                torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)

            optimizer.step()
            total_train_loss += loss.item()

        avg_train_loss = total_train_loss / max(1, len(train_loader))
        dev_res = evaluate(model, dev_loader, device, print_report=False)
        last_res = dev_res

        if best_res is None or dev_res["eval_f1"] > best_res["eval_f1"]:
            best_res = dev_res

        if print_each_epoch:
            print(
                f"[seed={seed}] epoch {epoch:02d} | train_loss={avg_train_loss:.4f} | "
                f"dev_f1={dev_res['eval_f1']:.4f} | dev_p={dev_res['eval_precision']:.4f} | dev_r={dev_res['eval_recall']:.4f}"
            )

    out = best_res if use_best_dev else last_res
    out = dict(out)
    out["seed"] = seed
    return out


# ============================================
# 10) Run N seeds and summarize
# ============================================
def run_n_iterations(
    n_runs: int = 10,
    seeds=None,
    num_epochs: int = 10,
    batch_size: int = 16,
    lr: float = 1e-3,
    use_best_dev: bool = True,
):
    if seeds is None:
        seeds = list(range(1, n_runs + 1))

    all_results = []
    for s in seeds:
        print("\n" + "=" * 90)
        print(f"RUN {len(all_results)+1}/{len(seeds)} (seed={s})")
        print("=" * 90)
        res = run_one_iteration(
            seed=s,
            num_epochs=num_epochs,
            batch_size=batch_size,
            lr=lr,
            use_best_dev=use_best_dev,
            print_each_epoch=True,
        )
        all_results.append(res)

    keys = ["eval_loss", "eval_precision", "eval_recall", "eval_f1", "eval_accuracy"]

    print("\n" + "#" * 90)
    print("Per-run results")
    print("#" * 90)
    print("seed\t" + "\t".join(keys))
    for r in all_results:
        print(
            f"{r['seed']}\t"
            f"{r['eval_loss']:.6f}\t"
            f"{r['eval_precision']:.6f}\t"
            f"{r['eval_recall']:.6f}\t"
            f"{r['eval_f1']:.6f}\t"
            f"{r['eval_accuracy']:.6f}"
        )

    print("\n" + "#" * 90)
    print("Average over runs (mean ± std)")
    print("#" * 90)
    for k in keys:
        vals = np.array([r[k] for r in all_results], dtype=float)
        mean = vals.mean()
        std = vals.std(ddof=1) if len(vals) > 1 else 0.0
        print(f"{k}: {mean:.6f} ± {std:.6f}")

    return all_results


# ============================================
# 11) Main
# ============================================
if __name__ == "__main__":
    # use_best_dev=True -> each seed reports best dev F1 achieved during training
    # set False -> each seed reports final epoch dev metrics
    run_n_iterations(
        n_runs=10,
        num_epochs=10,
        batch_size=16,
        lr=1e-3,
        use_best_dev=True,
    )
```

```python
"""
BiLSTM–Dynamic-Attention–LSTM on ACTI corpus (PyTorch, 10 seeds)
-----------------------------------------------------------------
- Uses ACTI-style files: tag.dic, train.txt, dev.txt
- Token-level sequence labelling with joint labels (BIOES-entity-relation-role)
- Word embedding (+ optional pre-trained vectors) + char-CNN + BiLSTM +
  multi-head self-attention + dynamic gating + GRU + LSTM decoder
- Evaluated with seqeval: Precision / Recall / F1 / Accuracy
- Runs N independent runs (different seeds), retraining from scratch each time
"""

import random
from collections import Counter
from typing import Dict, Optional, List, Tuple

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    classification_report,
)

# ============================================================
# 1) Paths and global hyperparameters
# ============================================================

TAG_DIC_PATH = "model/BiLSTM-dynamic-att-LSTM/data1/tag.dic"
TRAIN_PATH   = "model/BiLSTM-dynamic-att-LSTM/data1/train.txt"
DEV_PATH     = "model/BiLSTM-dynamic-att-LSTM/data1/dev.txt"

# Training hyperparameters (tune as needed)
NUM_EPOCHS   = 50
BATCH_SIZE   = 16
LEARNING_RATE = 5e-4
WEIGHT_DECAY  = 1e-4
GRAD_CLIP     = 5.0

# Whether to use best dev F1 across epochs per run
USE_BEST_DEV = True

# Optional: use pre-trained word embeddings
USE_PRETRAINED_EMB = False   # set True if you have word2vec vectors
EMB_DIM = 300                # must match vector dim if you use pre-trained

# ============================================================
# 2) Reproducibility helper
# ============================================================

def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# 3) Load labels
# ============================================================

def load_labels(path: str) -> List[str]:
    with open(path, encoding="utf-8") as f:
        labs = [line.strip() for line in f if line.strip()]
    return labs

labels = load_labels(TAG_DIC_PATH)
num_labels = len(labels)
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for i, lab in enumerate(labels)}

print("Number of labels:", num_labels)


# ============================================================
# 4) Load ACTI train/dev (tokens + labels)
#    Each line: "tok tok ...\tLAB LAB ..."
# ============================================================

def load_acti_file(path: str) -> Tuple[List[List[str]], List[List[str]]]:
    sentences, tags = [], []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue
            parts = line.split("\t")
            if len(parts) != 2:
                continue
            sent_str, tag_str = parts
            tokens = sent_str.split()
            labs = tag_str.split()
            if len(tokens) != len(labs):
                raise ValueError(f"Token/label mismatch in {path}")
            if len(tokens) == 0:
                continue
            sentences.append(tokens)
            tags.append(labs)
    return sentences, tags

train_tokens, train_labels = load_acti_file(TRAIN_PATH)
dev_tokens,   dev_labels   = load_acti_file(DEV_PATH)

print("Train sentences:", len(train_tokens))
print("Dev sentences:  ", len(dev_tokens))


# ============================================================
# 5) Build word & char vocab from training data
# ============================================================

word_counter = Counter()
char_counter = Counter()

for sent in train_tokens:
    word_counter.update(sent)
    for tok in sent:
        char_counter.update(list(tok))

word_vocab = {"<PAD>": 0, "<UNK>": 1}
for w in word_counter.keys():
    if w not in word_vocab:
        word_vocab[w] = len(word_vocab)

char_vocab = {"<PAD>": 0, "<UNK>": 1}
for ch in char_counter.keys():
    if ch not in char_vocab:
        char_vocab[ch] = len(char_vocab)

vocab_size = len(word_vocab)
char_vocab_size = len(char_vocab)
print("Word vocab size:", vocab_size)
print("Char vocab size:", char_vocab_size)


# ============================================================
# 6) Dataset + collate_fn (with chars)
# ============================================================

class ActiDataset(Dataset):
    def __init__(
        self,
        sentences: List[List[str]],
        labels: List[List[str]],
        word_vocab: Dict[str, int],
        char_vocab: Dict[str, int],
        label2id: Dict[str, int],
    ):
        self.sentences = sentences
        self.labels = labels
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.label2id = label2id

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx: int):
        tokens = self.sentences[idx]
        labs = self.labels[idx]

        token_ids = [self.word_vocab.get(t, self.word_vocab["<UNK>"]) for t in tokens]
        label_ids = [self.label2id[l] for l in labs]

        char_ids = []
        for tok in tokens:
            chars = [self.char_vocab.get(ch, self.char_vocab["<UNK>"]) for ch in tok]
            char_ids.append(torch.tensor(chars, dtype=torch.long))

        return {
            "token_ids": torch.tensor(token_ids, dtype=torch.long),
            "char_ids": char_ids,  # list[tensor(word_len)]
            "labels": torch.tensor(label_ids, dtype=torch.long),
        }


def make_collate_fn(
    word_vocab: Dict[str, int],
    char_vocab: Dict[str, int],
    pad_label_id: int = -1,
):
    pad_token_id = word_vocab["<PAD>"]
    pad_char_id = char_vocab["<PAD>"]

    def collate_fn(batch):
        token_seqs = [item["token_ids"] for item in batch]
        char_seqs  = [item["char_ids"] for item in batch]
        label_seqs = [item["labels"] for item in batch]

        lengths = [len(seq) for seq in token_seqs]
        max_len = max(lengths)

        max_word_len = 1
        for sent_chars in char_seqs:
            for wchars in sent_chars:
                if len(wchars) > max_word_len:
                    max_word_len = len(wchars)

        padded_tokens, padded_labels, padded_masks, padded_chars = [], [], [], []

        for tokens, sent_chars, labels_t in zip(token_seqs, char_seqs, label_seqs):
            seq_len = len(tokens)
            pad_len = max_len - seq_len

            tokens_padded = torch.cat(
                [tokens, torch.full((pad_len,), pad_token_id, dtype=torch.long)]
            )
            labels_padded = torch.cat(
                [labels_t, torch.full((pad_len,), pad_label_id, dtype=torch.long)]
            )

            mask = torch.cat(
                [
                    torch.ones(seq_len, dtype=torch.bool),
                    torch.zeros(pad_len, dtype=torch.bool),
                ]
            )

            word_char_tensors = []
            for i in range(max_len):
                if i < seq_len:
                    wchars = sent_chars[i]
                    wlen = len(wchars)
                    if wlen < max_word_len:
                        wchars = torch.cat(
                            [
                                wchars,
                                torch.full(
                                    (max_word_len - wlen,),
                                    pad_char_id,
                                    dtype=torch.long,
                                ),
                            ]
                        )
                    else:
                        wchars = wchars[:max_word_len]
                else:
                    wchars = torch.full((max_word_len,), pad_char_id, dtype=torch.long)
                word_char_tensors.append(wchars)

            char_tensor = torch.stack(word_char_tensors, dim=0)  # [T, C]

            padded_tokens.append(tokens_padded)
            padded_labels.append(labels_padded)
            padded_masks.append(mask)
            padded_chars.append(char_tensor)

        return {
            "input_ids": torch.stack(padded_tokens, dim=0),  # [B, T]
            "char_ids":  torch.stack(padded_chars,  dim=0),  # [B, T, C]
            "labels":    torch.stack(padded_labels, dim=0),  # [B, T]
            "mask":      torch.stack(padded_masks,  dim=0),  # [B, T]
        }

    return collate_fn


train_dataset = ActiDataset(train_tokens, train_labels, word_vocab, char_vocab, label2id)
dev_dataset   = ActiDataset(dev_tokens,   dev_labels,   word_vocab, char_vocab, label2id)


# ============================================================
# 7) Optional: load pre-trained word vectors (stub)
# ============================================================

def load_pretrained_vectors() -> Dict[str, np.ndarray]:
    """
    Placeholder: load your word2vec model here and return a dict:
        { "word": np.array([...], dtype=float32) }
    Make sure the vector dimension matches EMB_DIM.
    """
    # Example (pseudo-code):
    # from gensim.models import KeyedVectors
    # kv = KeyedVectors.load_word2vec_format("path/to/word2vec.bin", binary=True)
    # return {w: kv[w] for w in kv.key_to_index}
    return {}  # empty -> no pre-trained words found


def build_embedding_matrix(
    word_vocab: Dict[str, int],
    emb_dim: int,
    use_pretrained: bool = False,
) -> np.ndarray:
    matrix = np.random.normal(scale=0.1, size=(len(word_vocab), emb_dim)).astype("float32")
    if not use_pretrained:
        return matrix

    print("Loading pre-trained vectors...")
    pretrained = load_pretrained_vectors()
    found = 0
    for w, idx in word_vocab.items():
        if w in pretrained and pretrained[w].shape[0] == emb_dim:
            matrix[idx] = pretrained[w]
            found += 1
    print(f"Pre-trained vectors found for {found}/{len(word_vocab)} words.")
    return matrix


# ============================================================
# 8) BiLSTM–Dynamic-Attention–LSTM model
# ============================================================

class BiLSTMDynamicAttLSTM(nn.Module):
    def __init__(
        self,
        vocab_size: int,
        char_vocab_size: int,
        num_labels: int,
        word_emb_dim: int = 300,
        char_emb_dim: int = 15,
        cnn_filters: int = 15,
        cnn_kernel_size: int = 5,
        bilstm_dim: int = 300,
        lstm_dim: int = 600,
        num_heads: int = 4,
        dropout: float = 0.3,
        pretrained_word_emb: Optional[np.ndarray] = None,
    ):
        super().__init__()
        self.num_labels = num_labels
        self.cnn_filters = cnn_filters

        self.word_embedding = nn.Embedding(vocab_size, word_emb_dim, padding_idx=0)
        if pretrained_word_emb is not None:
            with torch.no_grad():
                self.word_embedding.weight.copy_(torch.from_numpy(pretrained_word_emb))

        self.char_embedding = nn.Embedding(char_vocab_size, char_emb_dim, padding_idx=0)

        self.char_cnn = nn.Conv1d(
            in_channels=char_emb_dim,
            out_channels=cnn_filters,
            kernel_size=cnn_kernel_size,
            padding=cnn_kernel_size // 2,
            bias=True,
        )

        self.encoder = nn.LSTM(
            input_size=word_emb_dim + cnn_filters,
            hidden_size=bilstm_dim // 2,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        self.attn = nn.MultiheadAttention(
            embed_dim=bilstm_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        self.W_s = nn.Linear(bilstm_dim * 2, bilstm_dim * 2)

        self.gru = nn.GRU(
            input_size=bilstm_dim * 2,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.decoder = nn.LSTM(
            input_size=lstm_dim,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(lstm_dim, num_labels)

        # bias_weight as in the paper
        self.bias_weight = nn.Parameter(torch.tensor(10.0))

    def forward(self, input_ids, char_ids, mask, labels=None):
        B, T = input_ids.shape
        _, _, C = char_ids.shape

        word_emb = self.word_embedding(input_ids)  # [B, T, D_w]

        # char-cnn
        char_ids_flat = char_ids.view(B * T, C)       # [B*T, C]
        char_emb = self.char_embedding(char_ids_flat) # [B*T, C, D_c]
        char_emb = char_emb.transpose(1, 2)           # [B*T, D_c, C]
        char_conv = torch.relu(self.char_cnn(char_emb))  # [B*T, F, C]
        char_rep = torch.max(char_conv, dim=2).values    # [B*T, F]
        char_rep = char_rep.view(B, T, self.cnn_filters) # [B, T, F]

        char_rep = char_rep * mask.unsqueeze(-1).float()

        enc_input = torch.cat([word_emb, char_rep], dim=-1)  # [B, T, D_w+F]
        enc_out, _ = self.encoder(enc_input)                 # [B, T, H]

        key_padding_mask = ~mask
        attn_out, _ = self.attn(enc_out, enc_out, enc_out, key_padding_mask=key_padding_mask)  # [B, T, H]

        concat_ha = torch.cat([enc_out, attn_out], dim=-1)   # [B, T, 2H]
        gamma = torch.sigmoid(self.W_s(concat_ha) + self.bias_weight)  # [B, T, 2H]
        xi = gamma * concat_ha                               # [B, T, 2H]

        gru_out, _ = self.gru(xi)                            # [B, T, lstm_dim]
        dec_out, _ = self.decoder(gru_out)                   # [B, T, lstm_dim]
        logits = self.classifier(self.dropout(dec_out))      # [B, T, num_labels]

        if labels is not None:
            loss_fn = nn.CrossEntropyLoss(ignore_index=-1)
            return loss_fn(logits.view(-1, self.num_labels), labels.view(-1))
        return logits.argmax(dim=-1)  # [B, T]


# ============================================================
# 9) Evaluation (seqeval)
# ============================================================

def evaluate(model, data_loader, device, print_report: bool = False):
    model.eval()
    seqeval_true, seqeval_pred = [], []
    total_loss, steps = 0.0, 0

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            loss = model(input_ids, char_ids, mask, labels=labels)
            total_loss += loss.item()
            steps += 1

            preds = model(input_ids, char_ids, mask, labels=None)  # [B, T]
            B, T = input_ids.shape

            for i in range(B):
                true_seq, pred_seq = [], []
                for j in range(T):
                    if not mask[i, j]:
                        continue
                    true_lab_id = int(labels[i, j].item())
                    if true_lab_id == -1:
                        continue
                    pred_lab_id = int(preds[i, j].item())
                    true_seq.append(id2label[true_lab_id])
                    pred_seq.append(id2label[pred_lab_id])

                if true_seq:
                    seqeval_true.append(true_seq)
                    seqeval_pred.append(pred_seq)

    avg_loss = total_loss / steps if steps > 0 else 0.0
    p = precision_score(seqeval_true, seqeval_pred)
    r = recall_score(seqeval_true, seqeval_pred)
    f = f1_score(seqeval_true, seqeval_pred)
    a = accuracy_score(seqeval_true, seqeval_pred)

    results = {
        "eval_loss": float(avg_loss),
        "eval_precision": float(p),
        "eval_recall": float(r),
        "eval_f1": float(f),
        "eval_accuracy": float(a),
    }

    if print_report:
        print("\nClassification report:\n")
        print(classification_report(seqeval_true, seqeval_pred))

    return results


# ============================================================
# 10) One run (one seed) training
# ============================================================

def run_one_iteration(
    seed: int,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LEARNING_RATE,
    weight_decay: float = WEIGHT_DECAY,
    grad_clip: float = GRAD_CLIP,
    use_best_dev: bool = True,
    print_each_epoch: bool = True,
):
    set_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    collate_fn = make_collate_fn(word_vocab, char_vocab, pad_label_id=-1)

    g = torch.Generator()
    g.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=collate_fn,
        generator=g,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )
    dev_loader = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=collate_fn,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    pretrained_matrix = build_embedding_matrix(
        word_vocab,
        emb_dim=EMB_DIM,
        use_pretrained=USE_PRETRAINED_EMB,
    )

    model = BiLSTMDynamicAttLSTM(
        vocab_size=vocab_size,
        char_vocab_size=char_vocab_size,
        num_labels=num_labels,
        word_emb_dim=EMB_DIM,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.3,
        pretrained_word_emb=pretrained_matrix,
    ).to(device)

    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=3,

    )

    best_res = None
    last_res = None

    for epoch in range(1, num_epochs + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            optimizer.zero_grad()
            loss = model(input_ids, char_ids, mask, labels=labels)
            loss.backward()

            if grad_clip is not None:
                torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)

            optimizer.step()
            total_train_loss += loss.item()

        avg_train_loss = total_train_loss / max(1, len(train_loader))
        dev_res = evaluate(model, dev_loader, device, print_report=False)
        last_res = dev_res

        scheduler.step(dev_res["eval_loss"])

        if best_res is None or dev_res["eval_f1"] > best_res["eval_f1"]:
            best_res = dev_res

        if print_each_epoch:
            print(
                f"[seed={seed}] epoch {epoch:02d} | "
                f"train_loss={avg_train_loss:.4f} | "
                f"dev_loss={dev_res['eval_loss']:.4f} | "
                f"dev_f1={dev_res['eval_f1']:.4f} | "
                f"dev_p={dev_res['eval_precision']:.4f} | "
                f"dev_r={dev_res['eval_recall']:.4f}"
            )

    out = best_res if use_best_dev else last_res
    out = dict(out)
    out["seed"] = seed
    return out


# ============================================================
# 11) Run N seeds and summarize
# ============================================================

def run_n_iterations(
    n_runs: int = 10,
    seeds: Optional[List[int]] = None,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LEARNING_RATE,
    use_best_dev: bool = True,
):
    if seeds is None:
        seeds = list(range(1, n_runs + 1))

    all_results = []
    for s in seeds:
        print("\n" + "=" * 90)
        print(f"RUN {len(all_results)+1}/{len(seeds)} (seed={s})")
        print("=" * 90)
        res = run_one_iteration(
            seed=s,
            num_epochs=num_epochs,
            batch_size=batch_size,
            lr=lr,
            use_best_dev=use_best_dev,
            print_each_epoch=True,
        )
        all_results.append(res)

    keys = ["eval_loss", "eval_precision", "eval_recall", "eval_f1", "eval_accuracy"]

    print("\n" + "#" * 90)
    print("Per-run results")
    print("#" * 90)
    print("seed\t" + "\t".join(keys))
    for r in all_results:
        print(
            f"{r['seed']}\t"
            f"{r['eval_loss']:.6f}\t"
            f"{r['eval_precision']:.6f}\t"
            f"{r['eval_recall']:.6f}\t"
            f"{r['eval_f1']:.6f}\t"
            f"{r['eval_accuracy']:.6f}"
        )

    print("\n" + "#" * 90)
    print("Average over runs (mean ± std)")
    print("#" * 90)
    for k in keys:
        vals = np.array([r[k] for r in all_results], dtype=float)
        mean = vals.mean()
        std = vals.std(ddof=1) if len(vals) > 1 else 0.0
        print(f"{k}: {mean:.6f} ± {std:.6f}")

    return all_results


# ============================================================
# 12) Main
# ============================================================

if __name__ == "__main__":
    run_n_iterations(
        n_runs=10,
        num_epochs=NUM_EPOCHS,
        batch_size=BATCH_SIZE,
        lr=LEARNING_RATE,
        use_best_dev=USE_BEST_DEV,
    )
```

```python
"""
BiLSTM–Dynamic-Attention–LSTM on ACTI (PyTorch, 10 seeds)
---------------------------------------------------------
- Uses ACTI-style tag.dic, train.txt, dev.txt
- Word + char-CNN + BiLSTM + self-attention + dynamic gate + GRU + LSTM
- Evaluation with seqeval (Precision / Recall / F1 / Accuracy)
- Evaluation collapses the 'role' part of the label (closer to ACTI paper)
- 10 independent runs with different seeds
"""

import random
from collections import Counter
from typing import Dict, Optional, List, Tuple

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    classification_report,
)

# ============================================================
# 1) Paths and global hyperparameters
# ============================================================

TAG_DIC_PATH = "model/BiLSTM-dynamic-att-LSTM/data1/tag.dic"
TRAIN_PATH   = "model/BiLSTM-dynamic-att-LSTM/data1/train.txt"
DEV_PATH     = "model/BiLSTM-dynamic-att-LSTM/data1/dev.txt"

# Training hyperparameters (closer to ACTI paper)
NUM_EPOCHS    = 125  # you can go to 125 if you want
BATCH_SIZE    = 16
LEARNING_RATE = 1e-3
WEIGHT_DECAY  = 0.0    # paper does not mention weight decay
GRAD_CLIP     = 5.0

USE_BEST_DEV = True    # keep best dev F1 over epochs

# Optional: pre-trained word embeddings (word2vec); off by default
USE_PRETRAINED_EMB = False
EMB_DIM = 300          # word embedding dimension

# Evaluation: whether to ignore the 'role' part of the label string
EVAL_IGNORE_ROLE = True

# ============================================================
# 2) Reproducibility helper
# ============================================================

def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# 3) Load labels
# ============================================================

def load_labels(path: str) -> List[str]:
    with open(path, encoding="utf-8") as f:
        labs = [line.strip() for line in f if line.strip()]
    return labs

labels = load_labels(TAG_DIC_PATH)
num_labels = len(labels)
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for i, lab in enumerate(labels)}

print("Number of labels:", num_labels)


# ============================================================
# 4) Load ACTI train/dev (tokens + labels)
#    Each line: "tok tok ...\tLAB LAB ..."
# ============================================================

def load_acti_file(path: str) -> Tuple[List[List[str]], List[List[str]]]:
    sentences, tags = [], []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue
            parts = line.split("\t")
            if len(parts) != 2:
                continue
            sent_str, tag_str = parts
            tokens = sent_str.split()
            labs = tag_str.split()
            if len(tokens) != len(labs):
                raise ValueError(f"Token/label mismatch in {path}")
            if len(tokens) == 0:
                continue
            sentences.append(tokens)
            tags.append(labs)
    return sentences, tags

train_tokens, train_labels = load_acti_file(TRAIN_PATH)
dev_tokens,   dev_labels   = load_acti_file(DEV_PATH)

print("Train sentences:", len(train_tokens))
print("Dev sentences:  ", len(dev_tokens))


# ============================================================
# 5) Build word & char vocab from training data
# ============================================================

word_counter = Counter()
char_counter = Counter()

for sent in train_tokens:
    word_counter.update(sent)
    for tok in sent:
        char_counter.update(list(tok))

word_vocab = {"<PAD>": 0, "<UNK>": 1}
for w in word_counter.keys():
    if w not in word_vocab:
        word_vocab[w] = len(word_vocab)

char_vocab = {"<PAD>": 0, "<UNK>": 1}
for ch in char_counter.keys():
    if ch not in char_vocab:
        char_vocab[ch] = len(char_vocab)

vocab_size = len(word_vocab)
char_vocab_size = len(char_vocab)
print("Word vocab size:", vocab_size)
print("Char vocab size:", char_vocab_size)


# ============================================================
# 6) Dataset + collate_fn (with chars)
# ============================================================

class ActiDataset(Dataset):
    def __init__(
        self,
        sentences: List[List[str]],
        labels: List[List[str]],
        word_vocab: Dict[str, int],
        char_vocab: Dict[str, int],
        label2id: Dict[str, int],
    ):
        self.sentences = sentences
        self.labels = labels
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.label2id = label2id

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx: int):
        tokens = self.sentences[idx]
        labs = self.labels[idx]

        token_ids = [self.word_vocab.get(t, self.word_vocab["<UNK>"]) for t in tokens]
        label_ids = [self.label2id[l] for l in labs]

        char_ids = []
        for tok in tokens:
            chars = [self.char_vocab.get(ch, self.char_vocab["<UNK>"]) for ch in tok]
            char_ids.append(torch.tensor(chars, dtype=torch.long))

        return {
            "token_ids": torch.tensor(token_ids, dtype=torch.long),
            "char_ids": char_ids,  # list[tensor(word_len)]
            "labels": torch.tensor(label_ids, dtype=torch.long),
        }


def make_collate_fn(
    word_vocab: Dict[str, int],
    char_vocab: Dict[str, int],
    pad_label_id: int = -1,
):
    pad_token_id = word_vocab["<PAD>"]
    pad_char_id = char_vocab["<PAD>"]

    def collate_fn(batch):
        token_seqs = [item["token_ids"] for item in batch]
        char_seqs  = [item["char_ids"] for item in batch]
        label_seqs = [item["labels"] for item in batch]

        lengths = [len(seq) for seq in token_seqs]
        max_len = max(lengths)

        max_word_len = 1
        for sent_chars in char_seqs:
            for wchars in sent_chars:
                if len(wchars) > max_word_len:
                    max_word_len = len(wchars)

        padded_tokens, padded_labels, padded_masks, padded_chars = [], [], [], []

        for tokens, sent_chars, labels_t in zip(token_seqs, char_seqs, label_seqs):
            seq_len = len(tokens)
            pad_len = max_len - seq_len

            tokens_padded = torch.cat(
                [tokens, torch.full((pad_len,), pad_token_id, dtype=torch.long)]
            )
            labels_padded = torch.cat(
                [labels_t, torch.full((pad_len,), pad_label_id, dtype=torch.long)]
            )

            mask = torch.cat(
                [
                    torch.ones(seq_len, dtype=torch.bool),
                    torch.zeros(pad_len, dtype=torch.bool),
                ]
            )

            word_char_tensors = []
            for i in range(max_len):
                if i < seq_len:
                    wchars = sent_chars[i]
                    wlen = len(wchars)
                    if wlen < max_word_len:
                        wchars = torch.cat(
                            [
                                wchars,
                                torch.full(
                                    (max_word_len - wlen,),
                                    pad_char_id,
                                    dtype=torch.long,
                                ),
                            ]
                        )
                    else:
                        wchars = wchars[:max_word_len]
                else:
                    wchars = torch.full((max_word_len,), pad_char_id, dtype=torch.long)
                word_char_tensors.append(wchars)

            char_tensor = torch.stack(word_char_tensors, dim=0)  # [T, C]

            padded_tokens.append(tokens_padded)
            padded_labels.append(labels_padded)
            padded_masks.append(mask)
            padded_chars.append(char_tensor)

        return {
            "input_ids": torch.stack(padded_tokens, dim=0),  # [B, T]
            "char_ids":  torch.stack(padded_chars,  dim=0),  # [B, T, C]
            "labels":    torch.stack(padded_labels, dim=0),  # [B, T]
            "mask":      torch.stack(padded_masks,  dim=0),  # [B, T]
        }

    return collate_fn


train_dataset = ActiDataset(train_tokens, train_labels, word_vocab, char_vocab, label2id)
dev_dataset   = ActiDataset(dev_tokens,   dev_labels,   word_vocab, char_vocab, label2id)


# ============================================================
# 7) Optional: load pre-trained word vectors (stub)
# ============================================================

def load_pretrained_vectors() -> Dict[str, np.ndarray]:
    """
    Placeholder: load your word2vec model here and return a dict:
        { "word": np.array([...], dtype=float32) }
    Make sure the vector dimension matches EMB_DIM.
    """
    # Example (pseudo-code):
    # from gensim.models import KeyedVectors
    # kv = KeyedVectors.load_word2vec_format("path/to/word2vec.bin", binary=True)
    # return {w: kv[w] for w in kv.key_to_index}
    return {}  # empty -> no pre-trained words found


def build_embedding_matrix(
    word_vocab: Dict[str, int],
    emb_dim: int,
    use_pretrained: bool = False,
) -> np.ndarray:
    matrix = np.random.normal(scale=0.1, size=(len(word_vocab), emb_dim)).astype("float32")
    if not use_pretrained:
        return matrix

    print("Loading pre-trained vectors...")
    pretrained = load_pretrained_vectors()
    found = 0
    for w, idx in word_vocab.items():
        if w in pretrained and pretrained[w].shape[0] == emb_dim:
            matrix[idx] = pretrained[w]
            found += 1
    print(f"Pre-trained vectors found for {found}/{len(word_vocab)} words.")
    return matrix


# ============================================================
# 8) BiLSTM–Dynamic-Attention–LSTM model
# ============================================================

class BiLSTMDynamicAttLSTM(nn.Module):
    def __init__(
        self,
        vocab_size: int,
        char_vocab_size: int,
        num_labels: int,
        word_emb_dim: int = 300,
        char_emb_dim: int = 15,
        cnn_filters: int = 15,
        cnn_kernel_size: int = 5,
        bilstm_dim: int = 300,
        lstm_dim: int = 600,
        num_heads: int = 4,
        dropout: float = 0.1,
        pretrained_word_emb: Optional[np.ndarray] = None,
    ):
        super().__init__()
        self.num_labels = num_labels
        self.cnn_filters = cnn_filters

        self.word_embedding = nn.Embedding(vocab_size, word_emb_dim, padding_idx=0)
        if pretrained_word_emb is not None:
            with torch.no_grad():
                self.word_embedding.weight.copy_(torch.from_numpy(pretrained_word_emb))

        self.char_embedding = nn.Embedding(char_vocab_size, char_emb_dim, padding_idx=0)

        self.char_cnn = nn.Conv1d(
            in_channels=char_emb_dim,
            out_channels=cnn_filters,
            kernel_size=cnn_kernel_size,
            padding=cnn_kernel_size // 2,
            bias=True,
        )

        self.encoder = nn.LSTM(
            input_size=word_emb_dim + cnn_filters,
            hidden_size=bilstm_dim // 2,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        self.attn = nn.MultiheadAttention(
            embed_dim=bilstm_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        self.W_s = nn.Linear(bilstm_dim * 2, bilstm_dim * 2)

        self.gru = nn.GRU(
            input_size=bilstm_dim * 2,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.decoder = nn.LSTM(
            input_size=lstm_dim,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(lstm_dim, num_labels)

        self.bias_weight = nn.Parameter(torch.tensor(10.0))

    def forward(self, input_ids, char_ids, mask, labels=None):
        B, T = input_ids.shape
        _, _, C = char_ids.shape

        word_emb = self.word_embedding(input_ids)  # [B, T, D_w]

        # char-cnn
        char_ids_flat = char_ids.view(B * T, C)       # [B*T, C]
        char_emb = self.char_embedding(char_ids_flat) # [B*T, C, D_c]
        char_emb = char_emb.transpose(1, 2)           # [B*T, D_c, C]
        char_conv = torch.relu(self.char_cnn(char_emb))  # [B*T, F, C]
        char_rep = torch.max(char_conv, dim=2).values    # [B*T, F]
        char_rep = char_rep.view(B, T, self.cnn_filters) # [B, T, F]

        char_rep = char_rep * mask.unsqueeze(-1).float()

        enc_input = torch.cat([word_emb, char_rep], dim=-1)  # [B, T, D_w+F]
        enc_out, _ = self.encoder(enc_input)                 # [B, T, H]

        key_padding_mask = ~mask
        attn_out, _ = self.attn(enc_out, enc_out, enc_out, key_padding_mask=key_padding_mask)  # [B, T, H]

        concat_ha = torch.cat([enc_out, attn_out], dim=-1)   # [B, T, 2H]
        gamma = torch.sigmoid(self.W_s(concat_ha) + self.bias_weight)  # [B, T, 2H]
        xi = gamma * concat_ha                               # [B, T, 2H]

        gru_out, _ = self.gru(xi)                            # [B, T, lstm_dim]
        dec_out, _ = self.decoder(gru_out)                   # [B, T, lstm_dim]
        logits = self.classifier(self.dropout(dec_out))      # [B, T, num_labels]

        if labels is not None:
            loss_fn = nn.CrossEntropyLoss(ignore_index=-1)
            return loss_fn(logits.view(-1, self.num_labels), labels.view(-1))
        return logits.argmax(dim=-1)  # [B, T]


# ============================================================
# 9) Label collapsing for evaluation (ignore role)
# ============================================================

def collapse_label(label_str: str) -> str:
    """
    Collapse BIOES-entity-relation-role -> BIOES-entity-relation
    so we ignore the 'role' part (closer to ACTI's evaluation).
    """
    if not EVAL_IGNORE_ROLE:
        return label_str
    if label_str == "O":
        return "O"
    parts = label_str.split("-")
    # Typically: [BIOES, entity, relation, role]
    if len(parts) <= 2:
        return label_str
    # drop the last part (role)
    return "-".join(parts[:-1])


# ============================================================
# 10) Evaluation (seqeval)
# ============================================================

def evaluate(model, data_loader, device, print_report: bool = False):
    model.eval()
    seqeval_true, seqeval_pred = [], []
    total_loss, steps = 0.0, 0

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            loss = model(input_ids, char_ids, mask, labels=labels)
            total_loss += loss.item()
            steps += 1

            preds = model(input_ids, char_ids, mask, labels=None)  # [B, T]
            B, T = input_ids.shape

            for i in range(B):
                true_seq, pred_seq = [], []
                for j in range(T):
                    if not mask[i, j]:
                        continue
                    true_lab_id = int(labels[i, j].item())
                    if true_lab_id == -1:
                        continue
                    pred_lab_id = int(preds[i, j].item())

                    true_lab = collapse_label(id2label[true_lab_id])
                    pred_lab = collapse_label(id2label[pred_lab_id])

                    true_seq.append(true_lab)
                    pred_seq.append(pred_lab)

                if true_seq:
                    seqeval_true.append(true_seq)
                    seqeval_pred.append(pred_seq)

    avg_loss = total_loss / steps if steps > 0 else 0.0
    p = precision_score(seqeval_true, seqeval_pred)
    r = recall_score(seqeval_true, seqeval_pred)
    f = f1_score(seqeval_true, seqeval_pred)
    a = accuracy_score(seqeval_true, seqeval_pred)

    results = {
        "eval_loss": float(avg_loss),
        "eval_precision": float(p),
        "eval_recall": float(r),
        "eval_f1": float(f),
        "eval_accuracy": float(a),
    }

    if print_report:
        print("\nClassification report:\n")
        print(classification_report(seqeval_true, seqeval_pred))

    return results


# ============================================================
# 11) One run (one seed) training
# ============================================================

def run_one_iteration(
    seed: int,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LEARNING_RATE,
    weight_decay: float = WEIGHT_DECAY,
    grad_clip: float = GRAD_CLIP,
    use_best_dev: bool = True,
    print_each_epoch: bool = True,
):
    set_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    collate_fn = make_collate_fn(word_vocab, char_vocab, pad_label_id=-1)

    g = torch.Generator()
    g.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=collate_fn,
        generator=g,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )
    dev_loader = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=collate_fn,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    pretrained_matrix = build_embedding_matrix(
        word_vocab,
        emb_dim=EMB_DIM,
        use_pretrained=USE_PRETRAINED_EMB,
    )

    model = BiLSTMDynamicAttLSTM(
        vocab_size=vocab_size,
        char_vocab_size=char_vocab_size,
        num_labels=num_labels,
        word_emb_dim=EMB_DIM,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,  # as in paper
        pretrained_word_emb=pretrained_matrix,
    ).to(device)

    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=3,
    )

    best_res = None
    last_res = None

    for epoch in range(1, num_epochs + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            optimizer.zero_grad()
            loss = model(input_ids, char_ids, mask, labels=labels)
            loss.backward()

            if grad_clip is not None:
                torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)

            optimizer.step()
            total_train_loss += loss.item()

        avg_train_loss = total_train_loss / max(1, len(train_loader))
        dev_res = evaluate(model, dev_loader, device, print_report=False)
        last_res = dev_res

        scheduler.step(dev_res["eval_loss"])

        if best_res is None or dev_res["eval_f1"] > best_res["eval_f1"]:
            best_res = dev_res

        if print_each_epoch:
            print(
                f"[seed={seed}] epoch {epoch:02d} | "
                f"train_loss={avg_train_loss:.4f} | "
                f"dev_loss={dev_res['eval_loss']:.4f} | "
                f"dev_f1={dev_res['eval_f1']:.4f} | "
                f"dev_p={dev_res['eval_precision']:.4f} | "
                f"dev_r={dev_res['eval_recall']:.4f}"
            )

    out = best_res if use_best_dev else last_res
    out = dict(out)
    out["seed"] = seed
    return out


# ============================================================
# 12) Run N seeds and summarize
# ============================================================

def run_n_iterations(
    n_runs: int = 10,
    seeds: Optional[List[int]] = None,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LEARNING_RATE,
    use_best_dev: bool = True,
):
    if seeds is None:
        seeds = list(range(1, n_runs + 1))

    all_results = []
    for s in seeds:
        print("\n" + "=" * 90)
        print(f"RUN {len(all_results)+1}/{len(seeds)} (seed={s})")
        print("=" * 90)
        res = run_one_iteration(
            seed=s,
            num_epochs=num_epochs,
            batch_size=batch_size,
            lr=lr,
            use_best_dev=use_best_dev,
            print_each_epoch=True,
        )
        all_results.append(res)

    keys = ["eval_loss", "eval_precision", "eval_recall", "eval_f1", "eval_accuracy"]

    print("\n" + "#" * 90)
    print("Per-run results")
    print("#" * 90)
    print("seed\t" + "\t".join(keys))
    for r in all_results:
        print(
            f"{r['seed']}\t"
            f"{r['eval_loss']:.6f}\t"
            f"{r['eval_precision']:.6f}\t"
            f"{r['eval_recall']:.6f}\t"
            f"{r['eval_f1']:.6f}\t"
            f"{r['eval_accuracy']:.6f}"
        )

    print("\n" + "#" * 90)
    print("Average over runs (mean ± std)")
    print("#" * 90)
    for k in keys:
        vals = np.array([r[k] for r in all_results], dtype=float)
        mean = vals.mean()
        std = vals.std(ddof=1) if len(vals) > 1 else 0.0
        print(f"{k}: {mean:.6f} ± {std:.6f}")

    return all_results


# ============================================================
# 13) Main
# ============================================================

if __name__ == "__main__":
    run_n_iterations(
        n_runs=10,
        num_epochs=NUM_EPOCHS,
        batch_size=BATCH_SIZE,
        lr=LEARNING_RATE,
        use_best_dev=USE_BEST_DEV,
    )
```

```python
"""
BiLSTM–Dynamic–Attention–LSTM on ACTI (PyTorch, improved training)

- Uses: tag.dic, train.txt, dev.txt in ACTI joint format
  Each line: "tok1 tok2 ...\tLAB1 LAB2 ..."
- Architecture: word+char embeddings -> BiLSTM -> Multihead self-attention
               -> dynamic gating -> GRU -> LSTM -> linear softmax
- Loss: token-level cross-entropy with ignore_index=-1
- Optimizer: Adam + ReduceLROnPlateau on dev F1
- Training: 10 seeds, 100 epochs (best dev F1 per seed)

You MUST set TAG_DIC_PATH, TRAIN_PATH, DEV_PATH to your own files.
"""

import random
from collections import Counter

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from seqeval.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    classification_report,
)

# ============================================
# 0) Paths + global hyperparameters
# ============================================

# TODO: change these to your local ACTI paths
TAG_DIC_PATH = "model/BiLSTM-dynamic-att-LSTM/data1/tag.dic"
TRAIN_PATH   = "model/BiLSTM-dynamic-att-LSTM/data1/train.txt"
DEV_PATH     = "model/BiLSTM-dynamic-att-LSTM/data1/dev.txt"

NUM_EPOCHS    = 125  # you can increase to 125 if still improving
BATCH_SIZE    = 16
LR            = 1e-3
WEIGHT_DECAY  = 0.0   # keep 0.0 to stay close to plain setup
DROPOUT       = 0.1
GRAD_CLIP     = 5.0
PAD_LABEL_ID  = -1


# ============================================
# 1) Reproducibility
# ============================================

def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================
# 2) Load labels / tag dictionary
# ============================================

def load_labels(path):
    with open(path, encoding="utf-8") as f:
        labs = [line.strip() for line in f if line.strip()]
    return labs

labels = load_labels(TAG_DIC_PATH)
num_labels = len(labels)
label2id = {lab: i for i, lab in enumerate(labels)}
id2label = {i: lab for i, lab in enumerate(labels)}

print("Number of labels:", num_labels)


# ============================================
# 3) Load ACTI train/dev (tokens + labels)
# ============================================

def load_acti_file(path):
    sentences, tags = [], []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue
            parts = line.split("\t")
            if len(parts) != 2:
                continue
            sent_str, tag_str = parts
            tokens = sent_str.split()
            labs   = tag_str.split()
            if len(tokens) != len(labs):
                raise ValueError(f"Token/label mismatch in {path}")
            if len(tokens) == 0:
                continue
            sentences.append(tokens)
            tags.append(labs)
    return sentences, tags

train_tokens, train_labels = load_acti_file(TRAIN_PATH)
dev_tokens,   dev_labels   = load_acti_file(DEV_PATH)

print("Train sentences:", len(train_tokens))
print("Dev sentences:  ", len(dev_tokens))


# ============================================
# 4) Build word & char vocab from training data
# ============================================

word_counter = Counter()
char_counter = Counter()

for sent in train_tokens:
    word_counter.update(sent)
    for tok in sent:
        char_counter.update(list(tok))

word_vocab = {"<PAD>": 0, "<UNK>": 1}
for w in word_counter.keys():
    if w not in word_vocab:
        word_vocab[w] = len(word_vocab)

char_vocab = {"<PAD>": 0, "<UNK>": 1}
for ch in char_counter.keys():
    if ch not in char_vocab:
        char_vocab[ch] = len(char_vocab)

vocab_size = len(word_vocab)
char_vocab_size = len(char_vocab)

print("Word vocab size:", vocab_size)
print("Char vocab size:", char_vocab_size)


# ============================================
# 5) Dataset + collate_fn
# ============================================

class ActiDataset(Dataset):
    def __init__(self, sentences, labels, word_vocab, char_vocab, label2id):
        self.sentences = sentences
        self.labels = labels
        self.word_vocab = word_vocab
        self.char_vocab = char_vocab
        self.label2id = label2id

    def __len__(self):
        return len(self.sentences)

    def __getitem__(self, idx):
        tokens = self.sentences[idx]
        labs   = self.labels[idx]

        token_ids = [self.word_vocab.get(t, self.word_vocab["<UNK>"]) for t in tokens]
        label_ids = [self.label2id[l] for l in labs]

        char_ids = []
        for tok in tokens:
            chars = [self.char_vocab.get(ch, self.char_vocab["<UNK>"]) for ch in tok]
            char_ids.append(torch.tensor(chars, dtype=torch.long))

        return {
            "token_ids": torch.tensor(token_ids, dtype=torch.long),
            "char_ids":  char_ids,  # list[tensor(word_len)]
            "labels":    torch.tensor(label_ids, dtype=torch.long),
        }


def make_collate_fn(word_vocab, char_vocab, pad_label_id=-1):
    pad_token_id = word_vocab["<PAD>"]
    pad_char_id  = char_vocab["<PAD>"]

    def collate_fn(batch):
        token_seqs = [item["token_ids"] for item in batch]
        char_seqs  = [item["char_ids"]  for item in batch]
        label_seqs = [item["labels"]    for item in batch]

        lengths = [len(seq) for seq in token_seqs]
        max_len = max(lengths)

        max_word_len = 1
        for sent_chars in char_seqs:
            for wchars in sent_chars:
                if len(wchars) > max_word_len:
                    max_word_len = len(wchars)

        padded_tokens, padded_labels, padded_masks, padded_chars = [], [], [], []

        for tokens, sent_chars, labels_t in zip(token_seqs, char_seqs, label_seqs):
            seq_len = len(tokens)
            pad_len = max_len - seq_len

            tokens_padded = torch.cat([
                tokens,
                torch.full((pad_len,), pad_token_id, dtype=torch.long),
            ])
            labels_padded = torch.cat([
                labels_t,
                torch.full((pad_len,), pad_label_id, dtype=torch.long),
            ])

            mask = torch.cat([
                torch.ones(seq_len, dtype=torch.bool),
                torch.zeros(pad_len, dtype=torch.bool),
            ])

            word_char_tensors = []
            for i in range(max_len):
                if i < seq_len:
                    wchars = sent_chars[i]
                    wlen = len(wchars)
                    if wlen < max_word_len:
                        wchars = torch.cat([
                            wchars,
                            torch.full((max_word_len - wlen,), pad_char_id, dtype=torch.long),
                        ])
                    else:
                        wchars = wchars[:max_word_len]
                else:
                    wchars = torch.full((max_word_len,), pad_char_id, dtype=torch.long)
                word_char_tensors.append(wchars)

            char_tensor = torch.stack(word_char_tensors, dim=0)  # [T, C]

            padded_tokens.append(tokens_padded)
            padded_labels.append(labels_padded)
            padded_masks.append(mask)
            padded_chars.append(char_tensor)

        return {
            "input_ids": torch.stack(padded_tokens, dim=0),  # [B, T]
            "char_ids":  torch.stack(padded_chars,  dim=0),  # [B, T, C]
            "labels":    torch.stack(padded_labels, dim=0),  # [B, T]
            "mask":      torch.stack(padded_masks,  dim=0),  # [B, T]
        }

    return collate_fn


train_dataset = ActiDataset(train_tokens, train_labels, word_vocab, char_vocab, label2id)
dev_dataset   = ActiDataset(dev_tokens,   dev_labels,   word_vocab, char_vocab, label2id)


# ============================================
# 6) BiLSTM–Dynamic-Attention–LSTM model
# ============================================

class BiLSTMDynamicAttLSTM(nn.Module):
    def __init__(
        self,
        vocab_size,
        char_vocab_size,
        num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=0.1,
    ):
        super().__init__()
        self.num_labels   = num_labels
        self.word_emb_dim = word_emb_dim
        self.char_emb_dim = char_emb_dim
        self.cnn_filters  = cnn_filters
        self.bilstm_dim   = bilstm_dim
        self.lstm_dim     = lstm_dim

        # word embeddings
        self.word_embedding = nn.Embedding(vocab_size, word_emb_dim, padding_idx=0)
        # char embeddings
        self.char_embedding = nn.Embedding(char_vocab_size, char_emb_dim, padding_idx=0)

        # char CNN
        self.char_cnn = nn.Conv1d(
            in_channels=char_emb_dim,
            out_channels=cnn_filters,
            kernel_size=cnn_kernel_size,
            padding=cnn_kernel_size // 2,
            bias=True,
        )

        # BiLSTM encoder
        self.encoder = nn.LSTM(
            input_size=word_emb_dim + cnn_filters,
            hidden_size=bilstm_dim // 2,
            num_layers=1,
            batch_first=True,
            bidirectional=True,
        )

        # Multi-head self-attention
        self.attn = nn.MultiheadAttention(
            embed_dim=bilstm_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        # LayerNorms for stability
        self.ln_enc = nn.LayerNorm(bilstm_dim)
        self.ln_att = nn.LayerNorm(bilstm_dim)
        self.ln_xi  = nn.LayerNorm(bilstm_dim * 2)

        # Dynamic attention gating
        self.W_s = nn.Linear(bilstm_dim * 2, bilstm_dim * 2)
        # bias weight (set to 0.0 to avoid immediate saturation)
        self.bias_weight = nn.Parameter(torch.tensor(0.0))

        # GRU
        self.gru = nn.GRU(
            input_size=bilstm_dim * 2,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        # Decoder LSTM
        self.decoder = nn.LSTM(
            input_size=lstm_dim,
            hidden_size=lstm_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=False,
        )

        self.dropout    = nn.Dropout(dropout)
        self.classifier = nn.Linear(lstm_dim, num_labels)

    def forward(self, input_ids, char_ids, mask, labels=None):
        """
        input_ids: [B, T]
        char_ids:  [B, T, C]
        mask:      [B, T] bool
        labels:    [B, T] or None
        """
        B, T = input_ids.shape
        _, _, C = char_ids.shape

        # word embeddings
        word_emb = self.word_embedding(input_ids)  # [B, T, D_w]

        # char CNN
        char_ids_flat = char_ids.view(B * T, C)               # [B*T, C]
        char_emb = self.char_embedding(char_ids_flat)         # [B*T, C, D_c]
        char_emb = char_emb.transpose(1, 2)                   # [B*T, D_c, C]
        char_conv = torch.relu(self.char_cnn(char_emb))       # [B*T, F, C]
        char_rep  = torch.max(char_conv, dim=2).values        # [B*T, F]
        char_rep  = char_rep.view(B, T, self.cnn_filters)     # [B, T, F]

        char_rep = char_rep * mask.unsqueeze(-1).float()

        # BiLSTM encoder
        enc_input = torch.cat([word_emb, char_rep], dim=-1)   # [B, T, D_w+F]
        enc_out, _ = self.encoder(enc_input)                  # [B, T, H]
        enc_out = self.ln_enc(enc_out)

        # self-attention
        key_padding_mask = ~mask
        attn_out, _ = self.attn(enc_out, enc_out, enc_out, key_padding_mask=key_padding_mask)  # [B, T, H]
        attn_out = self.ln_att(attn_out)

        # dynamic attention gating
        concat_ha = torch.cat([enc_out, attn_out], dim=-1)    # [B, T, 2H]
        concat_ha = self.ln_xi(concat_ha)
        gamma = torch.sigmoid(self.W_s(concat_ha) + self.bias_weight)  # [B, T, 2H]
        xi = gamma * concat_ha                                # [B, T, 2H]

        # GRU + decoder
        gru_out, _ = self.gru(xi)                             # [B, T, lstm_dim]
        dec_out, _ = self.decoder(gru_out)                    # [B, T, lstm_dim]

        logits = self.classifier(self.dropout(dec_out))       # [B, T, num_labels]

        if labels is not None:
            loss_fn = nn.CrossEntropyLoss(ignore_index=PAD_LABEL_ID)
            return loss_fn(logits.view(-1, self.num_labels), labels.view(-1))
        else:
            return logits.argmax(dim=-1)                      # [B, T]


# ============================================
# 7) Evaluation with seqeval
# ============================================

def evaluate(model, data_loader, device, print_report=False):
    model.eval()
    seqeval_true, seqeval_pred = [], []
    total_loss, steps = 0.0, 0

    with torch.no_grad():
        for batch in data_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            loss = model(input_ids, char_ids, mask, labels=labels)
            total_loss += loss.item()
            steps += 1

            preds = model(input_ids, char_ids, mask, labels=None)  # [B, T]
            B, T = input_ids.shape

            for i in range(B):
                true_seq, pred_seq = [], []
                for j in range(T):
                    if not mask[i, j]:
                        continue
                    true_lab_id = int(labels[i, j].item())
                    if true_lab_id == PAD_LABEL_ID:
                        continue
                    pred_lab_id = int(preds[i, j].item())
                    true_seq.append(id2label[true_lab_id])
                    pred_seq.append(id2label[pred_lab_id])

                if true_seq:
                    seqeval_true.append(true_seq)
                    seqeval_pred.append(pred_seq)

    avg_loss = total_loss / steps if steps > 0 else 0.0
    p = precision_score(seqeval_true, seqeval_pred)
    r = recall_score(seqeval_true, seqeval_pred)
    f = f1_score(seqeval_true, seqeval_pred)
    a = accuracy_score(seqeval_true, seqeval_pred)

    results = {
        "eval_loss": float(avg_loss),
        "eval_precision": float(p),
        "eval_recall": float(r),
        "eval_f1": float(f),
        "eval_accuracy": float(a),
    }

    if print_report:
        print("\nClassification report:\n")
        print(classification_report(seqeval_true, seqeval_pred))

    return results


# ============================================
# 8) One run (one seed)
# ============================================

def run_one_seed(
    seed: int,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LR,
    weight_decay: float = WEIGHT_DECAY,
    grad_clip: float = GRAD_CLIP,
    use_best_dev: bool = True,
):
    set_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"\nUsing device {device}, seed={seed}")

    collate_fn = make_collate_fn(word_vocab, char_vocab, pad_label_id=PAD_LABEL_ID)

    g = torch.Generator()
    g.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        collate_fn=collate_fn,
        generator=g,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )
    dev_loader = DataLoader(
        dev_dataset,
        batch_size=batch_size,
        shuffle=False,
        collate_fn=collate_fn,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    model = BiLSTMDynamicAttLSTM(
        vocab_size=vocab_size,
        char_vocab_size=char_vocab_size,
        num_labels=num_labels,
        word_emb_dim=300,
        char_emb_dim=15,
        cnn_filters=15,
        cnn_kernel_size=5,
        bilstm_dim=300,
        lstm_dim=600,
        num_heads=4,
        dropout=DROPOUT,
    ).to(device)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=lr,
        weight_decay=weight_decay,
    )
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",       # we monitor dev F1 (higher is better)
        factor=0.5,
        patience=5,
        min_lr=1e-5,
    )

    best_res = None
    last_res = None

    for epoch in range(1, num_epochs + 1):
        model.train()
        total_train_loss = 0.0

        for batch in train_loader:
            input_ids = batch["input_ids"].to(device)
            char_ids  = batch["char_ids"].to(device)
            labels    = batch["labels"].to(device)
            mask      = batch["mask"].to(device)

            optimizer.zero_grad()
            loss = model(input_ids, char_ids, mask, labels=labels)
            loss.backward()

            if grad_clip is not None:
                nn.utils.clip_grad_norm_(model.parameters(), grad_clip)

            optimizer.step()
            total_train_loss += loss.item()

        avg_train_loss = total_train_loss / max(1, len(train_loader))
        dev_res = evaluate(model, dev_loader, device, print_report=False)
        last_res = dev_res

        scheduler.step(dev_res["eval_f1"])

        if best_res is None or dev_res["eval_f1"] > best_res["eval_f1"]:
            best_res = dev_res

        print(
            f"[seed={seed}] epoch {epoch:03d} | "
            f"train_loss={avg_train_loss:.4f} | "
            f"dev_f1={dev_res['eval_f1']:.4f} | "
            f"dev_p={dev_res['eval_precision']:.4f} | "
            f"dev_r={dev_res['eval_recall']:.4f}"
        )

    out = best_res if use_best_dev else last_res
    out = dict(out)
    out["seed"] = seed
    return out


# ============================================
# 9) Run N seeds and summarise
# ============================================

def run_n_seeds(
    n_runs: int = 10,
    seeds=None,
    num_epochs: int = NUM_EPOCHS,
    batch_size: int = BATCH_SIZE,
    lr: float = LR,
    use_best_dev: bool = True,
):
    if seeds is None:
        seeds = list(range(1, n_runs + 1))

    all_results = []
    for s in seeds:
        print("\n" + "=" * 90)
        print(f"RUN {len(all_results)+1}/{len(seeds)} (seed={s})")
        print("=" * 90)
        res = run_one_seed(
            seed=s,
            num_epochs=num_epochs,
            batch_size=batch_size,
            lr=lr,
            use_best_dev=use_best_dev,
        )
        all_results.append(res)

    keys = ["eval_loss", "eval_precision", "eval_recall", "eval_f1", "eval_accuracy"]

    print("\n" + "#" * 90)
    print("Per-run results")
    print("#" * 90)
    print("seed\t" + "\t".join(keys))
    for r in all_results:
        print(
            f"{r['seed']}\t"
            f"{r['eval_loss']:.6f}\t"
            f"{r['eval_precision']:.6f}\t"
            f"{r['eval_recall']:.6f}\t"
            f"{r['eval_f1']:.6f}\t"
            f"{r['eval_accuracy']:.6f}"
        )

    print("\n" + "#" * 90)
    print("Average over runs (mean ± std)")
    print("#" * 90)
    for k in keys:
        vals = np.array([r[k] for r in all_results], dtype=float)
        mean = vals.mean()
        std  = vals.std(ddof=1) if len(vals) > 1 else 0.0
        print(f"{k}: {mean:.6f} ± {std:.6f}")

    return all_results


# ============================================
# 10) Main
# ============================================

if __name__ == "__main__":
    run_n_seeds(
        n_runs=10,
        num_epochs=NUM_EPOCHS,
        batch_size=BATCH_SIZE,
        lr=LR,
        use_best_dev=True,
    )
```
